**Homework 0: Numpy**

In the homework assignment we explore the Numpy python package. For a brief introduction, go [here](https://numpy.org/devdocs/user/absolute_beginners.html).

Begin by importing the Numpy library:

```python
import numpy as np
```

Create a numpy array called `X` with 100 rows and 10 columns, where each entry is the number 1.

```python
X=np.ones((100,10))
```

Create a numpy array with one row and 1000 columns called `y`, where the entries are the numbers 1 through 1000 (in order).

```python
y=np.arange(1,1001)
```

```python
y[999]
```

Create a numpy array called `Y` with 100 rows and 10 columns, whose entries come from the array `y`.

```python
Y=y.reshape((100,10))
```

```python
Y.shape
```

Use the np.newaxis command to create an array called `z`, with enties from `y`, which has 1000 rows and one column.

```python
z=y[:,np.newaxis]
```

```python
z.shape
```

Create an array of shape (5,4) called `A`, consisting of those entries in `Y` in the first 5 rows and columns 2-5.  

```python
A=Y[0:5,2:6]
```

```python
A[3,3]
```

Create a 1-dimensional array called `b` consisting of the entries in `Y` that are divisible by 3, and between 20 and 70. (Do this by slicing `Y`, rather than creating the array from scratch.)


```python
b=Y[(Y%3==0) & (Y>20) & (Y<70)]
```

```python
b
```

Create an array called `Z` whose entries are the square-root of the entries in `Y`, plus the corresponding entries in `X`.

```python
Z=np.sqrt(Y)+X
```

```python
Z[5,5]
```

Create a 1-D array `m` whose entries are the maximum of each column of `Y`.

```python
m=Y.max(axis=0)
```

```python
m
```

Create a 1-D array `s` whose entries are the sums of each row of `Y`.

```python
s=Y.sum(axis=1)
```

```python
s[5]
```

Create an array `P` which is the product of `X` and the transpose of `Y`.

```python
P=np.matmul(X,Y.T)
```

```python
P[5,5]
```
