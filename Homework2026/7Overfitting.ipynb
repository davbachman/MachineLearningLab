**Homework 7: Overfitting**

We'll pick up where we left off in Homework 6, but this time import sklearn versions of all relevant ML classes:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

cars=pd.read_csv('https://vincentarelbundock.github.io/Rdatasets/csv/causaldata/auto.csv')
disp=np.array(cars.displacement)
mpg=np.array(cars.mpg)

index=np.argsort(disp)
disp=disp[index][:,np.newaxis]
mpg=mpg[index]
```

Use `train_test_split` to split the `disp` and `mpg` arrays into arrays called `Xtrain`, `Xtest`, `ytrain`, and `ytest` with an 80/20 train/test split.

```python
Xtrain,Xtest,ytrain,ytest=train_test_split(disp,mpg,test_size=0.8)
```

Fit a `StandardScaler` object to `Xtrain`, and use that object to scale both `Xtrain` and `Xtest`, producing arrays `scaled_Xtrain` and `scaled_Xtest`.  

```python
disp_scaler=StandardScaler()
disp_scaler.fit(Xtrain)
scaled_Xtrain=disp_scaler.transform(Xtrain)
scaled_Xtest=disp_scaler.transform(Xtest)
```

```python
scaled_Xtest[10]
```

Create an eigth degree polynomial model, and fit it to `scaled_Xtrain` and `ytrain`. Finally, use that model to make predictions based on both `scaled_Xtrain` and `scaled_Xtest`.

```python
deg8=PolynomialFeatures(8,include_bias=True)
deg8_train_feats=deg8.fit_transform(scaled_Xtrain)
deg8_test_feats=deg8.fit_transform(scaled_Xtest)

deg8_model=LinearRegression(fit_intercept=False)
deg8_model.fit(deg8_train_feats,ytrain)

deg8_train_predictions=deg8_model.predict(deg8_train_feats)
deg8_test_predictions=deg8_model.predict(deg8_test_feats)
```
